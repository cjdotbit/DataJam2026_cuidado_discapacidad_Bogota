## I.1 · Problemática

Cuando una persona con discapacidad requiere ayuda de otra persona para su vida diaria, alguien tiene que prestarla.
En Bogotá, ese alguien es abrumadoramente **una mujer de la familia**: sin remuneración, sin contrato, sin cotización a pensión y sin sustituto institucional.

De las **31.298** personas certificadas que declaran requerir ayuda de otra persona, el **85,5 %** es cuidada por una mujer de su familia — una razón de **5,9 mujeres por cada hombre cuidador**.

> Solo la madre aparece **17.020** veces. El padre aparece **1.206**. Esa diferencia de catorce a uno no es la particularidad de un barrio ni de un año, se repite en las veinte localidades y en las siete vigencias del registro.

El problema público no es que exista cuidado familiar. Es que ese cuidado opera como **un impuesto invisible cobrado exclusivamente a las mujeres**, que el Distrito no mide, no territorializa y por tanto no puede compensar ni redistribuir de manera focalizada.

### El cuidado recorre toda la vida de las mujeres

La base permite ver algo que un promedio agregado oculta, la identidad de quien cuida cambia con la etapa de vida de la persona con discapacidad, pero **su sexo no cambia nunca**.

| Momento del curso de vida | Cuidadora principal | Contraparte masculina |
|---|---|---|
| Primera infancia | Madre **88,7 %** | Padre 4,2 % |
| Infancia | Madre **86,6 %** | Padre 5,6 % |
| Adolescencia | Madre **83,8 %** | Padre 7,0 % |
| Juventud | Madre **84,9 %** | Padre 6,1 % |
| Adultez | Madre **42,2 %** · Esposa 14,9 % | Padre 3,1 % · Esposo 9,3 % |
| Persona mayor de 60 | Hija **38,4 %** · Esposa 21,3 % | Hijo 11,0 % · Esposo 8,8 % |



La lectura de esta tabla es la parte más contundente del diagnóstico:

- En la **infancia** cuida la madre.
- En la **adultez** de la persona con discapacidad **sigue cuidando la madre en cuatro de cada diez casos** — una mujer que ya envejeció cuidando y que nunca fue relevada.
- Cuando la persona con discapacidad es **mayor de sesenta**, quien cuida es la hija, con frecuencia **tres veces y media superior** a la del hijo; o la esposa, con frecuencia **dos veces y media superior** a la del esposo.

**El cuidado de la discapacidad no se reparte entre hombres y mujeres en ninguna etapa de la vida. Se transfiere de una mujer a otra.**

### Consecuencias que el Distrito no está midiendo

| | |
|---|---|
| **Costo de oportunidad no compensado** | Cuidar a tiempo completo es incompatible con el empleo formal, la educación y la cotización pensional. La mujer que cuida hoy será una persona mayor sin pensión mañana: la discapacidad de un familiar se convierte en determinante de la pobreza futura de otra persona. |
| **Ausencia de relevo** | El 42,2 % de las personas adultas con discapacidad siguen siendo cuidadas por su madre. Cuando esa madre ya no pueda, no hay relevo previsto ni institucional ni familiar. Es una demanda de servicios diferida que hoy no aparece en ninguna proyección. |
| **Invisibilidad estadística** | El registro identifica a la persona con discapacidad, no a quien la cuida. Sabemos el parentesco, pero no su edad, educación, situación laboral ni si ella misma requiere cuidado. La política tiene un sujeto que no puede caracterizar. |
| **Focalización ciega** | Sin una medida territorial de la carga de cuidado, la oferta del Sistema Distrital de Cuidado no puede verificarse contra la demanda que debería atender. |

## I.2 · Justificación

### Es un problema medible con datos ya publicados

Toda la evidencia de este cuaderno sale de fuentes públicas, sin acuerdos de acceso ni datos reservados. Eso importa por dos razones. De **viabilidad**: en una ventana de tres semanas, un ejercicio que depende de conseguir información no publicada no se termina. De **replicabilidad institucional**: si el Distrito puede reproducir el indicador con lo que ya publica, el producto sobrevive al DataJam.

### El enfoque de género es estructural, no decorativo

Los términos de referencia valoran la inclusión del enfoque de género. En la mayoría de ejercicios eso se resuelve desagregando resultados por sexo al final. Aquí el género **no es una desagregación: es el hallazgo**. La variable dependiente del análisis —quién presta el cuidado— tiene una distribución por sexo de 5,9 a 1, y esa distribución es el fenómeno que se quiere explicar y territorializar.

### Conecta dos políticas que hoy se miran poco

La política de discapacidad y la política de cuidado tienen sujetos superpuestos y rectorías distintas. La **persona con discapacidad** es competencia de la Secretaría Distrital de Integración Social; la **mujer que la cuida** es sujeto del Sistema Distrital de Cuidado, liderado por la Secretaría Distrital de la Mujer. El ejercicio produce el objeto analítico que ambas necesitan: una medida territorial de la demanda de cuidado por discapacidad, comparable con la oferta desplegada.

## A.1 · Descargar los datos


```python
from pathlib import Path
import urllib.request, urllib.error, json, ssl, time

# ============================================================================
# DESCARGADOR ROBUSTO — lo usan todas las secciones del cuaderno
# ============================================================================
UA = ("Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
      "(KHTML, like Gecko) Chrome/122.0 Safari/537.36")

def _abrir(url, timeout=120, inseguro=False):
    req = urllib.request.Request(url, headers={
        "User-Agent": UA,
        "Accept": "*/*",
        "Accept-Language": "es-CO,es;q=0.9",
    })
    ctx = ssl._create_unverified_context() if inseguro else None
    return urllib.request.urlopen(req, timeout=timeout, context=ctx)

def bajar(url, destino=None, intentos=3, timeout=120, verbose=True):
    """Descarga una URL. Devuelve la ruta si hay destino, o los bytes si no.
    Reintenta, y si el problema es el certificado lo reporta y reintenta sin verificar."""
    destino = Path(destino) if destino else None
    if destino and destino.exists() and destino.stat().st_size > 1000:
        if verbose: print(f"    (ya estaba en disco: {destino.name})")
        return destino
    ultimo = None
    for inseguro in (False, True):
        for k in range(intentos):
            try:
                with _abrir(url, timeout, inseguro) as r:
                    datos = r.read()
                if inseguro and verbose:
                    print("    ⚠ Se descargó SIN verificar el certificado TLS.")
                if destino:
                    destino.write_bytes(datos)
                    return destino
                return datos
            except urllib.error.HTTPError as e:
                ultimo = f"HTTP {e.code} {e.reason}"
                if e.code in (403, 404):     # no vale la pena reintentar
                    break
            except ssl.SSLError as e:
                ultimo = f"SSL: {e}"
                break                         # pasa al intento sin verificar
            except Exception as e:
                ultimo = f"{type(e).__name__}: {e}"
            time.sleep(2 * (k + 1))
    raise RuntimeError(f"No se pudo descargar.\n    URL: {url}\n    Último error: {ultimo}")

def bajar_json(url, **kw):
    return json.loads(bajar(url, **kw).decode("utf-8"))

# ============================================================================
# FUENTE 1 — Registro de discapacidad certificada
# ============================================================================
DATASET     = "personas-certificadas-con-discapacidad-en-bogota-d-c"
URL_DIRECTA = ("https://datosabiertos.bogota.gov.co/dataset/"
               "202c1e53-dad8-4307-b16c-d37ed8e510d5/resource/"
               "1a48e972-b838-4090-a3fb-48a48558484f/download/"
               "osb_demografia-discapacidad.csv")
API_CKAN    = f"https://datosabiertos.bogota.gov.co/api/3/action/package_show?id={DATASET}"

ARCHIVO  = Path("osb_demografia-discapacidad.csv")
ENCODING = "utf-8-sig"          # tolera BOM; 'utf-8' rompería la 1ª columna

try:
    from google.colab import files
    EN_COLAB = True
except ImportError:
    EN_COLAB = False

origen = None
try:
    bajar(URL_DIRECTA, ARCHIVO)
    origen = "enlace directo"
except Exception as e:
    print(f"  · Enlace directo falló:\n    {e}\n  · Consultando el catálogo CKAN…")
    try:
        meta = bajar_json(API_CKAN)
        url2 = next(r["url"] for r in meta["result"]["resources"]
                    if r.get("url","").lower().endswith(".csv")
                    and "metadato" not in r["url"].lower())
        bajar(url2, ARCHIVO)
        origen = "API CKAN"
    except Exception as e2:
        print(f"  · El catálogo tampoco respondió:\n    {e2}")
        if EN_COLAB:
            print("\n  Suba el archivo manualmente:")
            subidos = files.upload()
            ARCHIVO = Path(list(subidos.keys())[0])
            origen = "subida manual"
        else:
            raise SystemExit("Coloque osb_demografia-discapacidad.csv junto al cuaderno.")

print(f"  ✓ Origen: {origen}")
print(f"    {ARCHIVO}  ({ARCHIVO.stat().st_size/1e6:.1f} MB)")
```

### A.7.1 · Utilidades de vinculación

```python
import unicodedata, re

LOCALIDADES = {
    1: "Usaquén",        2: "Chapinero",        3: "Santa Fe",
    4: "San Cristóbal",  5: "Usme",             6: "Tunjuelito",
    7: "Bosa",           8: "Kennedy",          9: "Fontibón",
    10: "Engativá",     11: "Suba",            12: "Barrios Unidos",
    13: "Teusaquillo",  14: "Los Mártires",    15: "Antonio Nariño",
    16: "Puente Aranda",17: "La Candelaria",   18: "Rafael Uribe Uribe",
    19: "Ciudad Bolívar",20: "Sumapaz",
}

# Variantes reales encontradas en los conjuntos del portal distrital.
# Las tres últimas salieron del propio servicio de Manzanas del Cuidado.
ALIAS = {
    "RAFAEL URIBE": 18, "SAN CRISTOBAL SUR": 4, "CANDELARIA": 17,
    "MARTIRES": 14, "LOS MARTIRES": 14, "CIUDAD BOLIVAR": 19,
    "ANTONIO NARINO": 15, "USAQUEN": 1, "SANTAFE": 3,
    "FONTIBON": 9, "ENGATIVA": 10,
    "LOS MARTIRES ": 14,                       # 'Los Martíres' (tilde mal puesta)
    "SANTAFE LA CANDELARIA": 3,                # manzana compartida -> se asigna a Santa Fe
    "SANTA FE Y LA CANDELARIA": 3,             # idem, otra grafía
}

def normalizar(texto):
    """'Ciudad Bolívar ' -> 'CIUDAD BOLIVAR'. Quita tildes, puntuación y espacios repetidos."""
    if pd.isna(texto):
        return ""
    t = unicodedata.normalize("NFD", str(texto).strip().upper())
    t = "".join(ch for ch in t if unicodedata.category(ch) != "Mn")
    return re.sub(r"\s+", " ", re.sub(r"[^A-Z0-9 ]", " ", t)).strip()

_INDICE = {normalizar(v): k for k, v in LOCALIDADES.items()}
_INDICE.update({normalizar(k): v for k, v in ALIAS.items()})

def codigo_localidad(valor):
    """Devuelve el código 1-20 a partir de un NOMBRE o de un CÓDIGO.

    El servicio REST de Manzanas del Cuidado devuelve el campo LOCALIDAD unas
    veces como nombre ('Ciudad Bolívar') y otras como código con cero a la
    izquierda ('19'), según cómo resuelva el dominio de valores codificados.
    Esta función acepta las dos formas."""
    if pd.isna(valor):
        return pd.NA
    t = str(valor).strip()
    # ¿Es un código? '19', '07', 19, 19.0
    try:
        n = int(float(t))
        if 1 <= n <= 20 and len(t.replace(".0", "")) <= 2:
            return n
    except (ValueError, TypeError):
        pass
    return _INDICE.get(normalizar(t), pd.NA)

def agregar_codigo(tabla, col_nombre, col_codigo="COD_LOC"):
    out = tabla.copy()
    out[col_codigo] = out[col_nombre].apply(codigo_localidad)
    sin = out[out[col_codigo].isna()][col_nombre].dropna().unique()
    if len(sin):
        print(f"  ⚠ {len(sin)} valor(es) sin código: {list(sin)[:8]}"
              + (" ..." if len(sin) > 8 else ""))
        print("    Si son nombres, agréguelos al diccionario ALIAS.")
        print("    Si son códigos fuera del rango 1-20, revise la fuente.")
    else:
        print(f"  ✓ {out[col_codigo].notna().sum()} filas con código asignado.")
    return out

def vincular(izq, der, sobre, nombre_izq="base", nombre_der="externa", como="left"):
    """Une por código y REPORTA lo que no cruzó en ambos sentidos."""
    sobre = list(sobre)
    dup = der.duplicated(subset=sobre).sum()
    if dup:
        print(f"  ⚠ {nombre_der}: {dup} filas duplicadas en la llave {sobre}.")
        print("    El merge multiplicaría registros. Agregue antes de vincular.")
    unido = izq.merge(der, on=sobre, how=como, indicator=True)
    solo_izq = unido[unido["_merge"] == "left_only"]
    k_izq = set(map(tuple, izq[sobre].dropna().values))
    k_der = set(map(tuple, der[sobre].dropna().values))
    print(f"  {nombre_izq} ({len(izq)}) ⨝ {nombre_der} ({len(der)}) -> {len(unido)} filas")
    if len(unido) != len(izq) and como == "left":
        print("  ⚠ El resultado cambió de tamaño: revise duplicados en la llave.")
    if len(solo_izq):
        print(f"  ⚠ {len(solo_izq)} filas de {nombre_izq} sin pareja:")
        for _, r in solo_izq[sobre].drop_duplicates().head(10).iterrows():
            print(f"      código {r[sobre[0]]} = {LOCALIDADES.get(r[sobre[0]], '?')}")
    sobra = k_der - k_izq
    if sobra:
        muestra = sorted(sobra)[:6]
        print(f"  · {len(sobra)} llaves presentes solo en {nombre_der} "
              f"(no es error si la fuente cubre más años): {muestra}"
              + (" ..." if len(sobra) > 6 else ""))
    if not len(solo_izq) and not (k_der - k_izq):
        print(f"  ✓ Cruce completo: las {len(k_izq)} llaves encontraron pareja.")
    return unido.drop(columns="_merge")

def verificar_cobertura(tabla, col="COD_LOC"):
    presentes = {int(c) for c in tabla[col].dropna().unique()}
    faltan = set(LOCALIDADES) - presentes
    print(f"  ⚠ Faltan: {sorted(faltan)}" if faltan else "  ✓ Cobertura completa: 20 localidades.")
    return not faltan

for t in ["Ciudad Bolívar", "RAFAEL  URIBE ", "Los Martíres", "Santa Fe y La Candelaria"]:
    print(f"  {t!r:28} -> código {codigo_localidad(t)}")
```

## R2 · ¿En qué localidades la feminización del cuidado es mayor?

*Subpregunta 2.*

Antes de comparar territorios hay que resolver una objeción: un territorio que certifica muchos niños tendrá feminización alta **solo porque en la infancia cuida la madre**. La estandarización aplica a cada localidad la mezcla de etapas de vida del conjunto de la ciudad; lo que quede después no puede atribuirse a composición.

Se retiró el índice compuesto ICCN que ocupaba este lugar: su ranking lo encabezaban las localidades más pequeñas y reordenaba la ciudad cada año.

```python
# ---------- Indicadores crudos por localidad --------------------------------
ind_loc = cui.groupby("NOMBRE_LOCALIDAD").apply(indicadores, include_groups=False)
base_dep = dep[~dep["TRUNCADO"]]
ind_loc = ind_loc.join(
    base_dep.groupby("NOMBRE_LOCALIDAD").apply(lambda x: pd.Series({
        "certificados": len(x),
        "DC": (x["REQUIEREAYUDADEOTRAPERSONA"] == "Si").mean() * 100,
    }), include_groups=False))

# ---------- IFC estandarizada por estructura etaria -------------------------
# Estandarización directa: a cada localidad se le aplica la mezcla de etapas
# de vida de toda la ciudad. Si el gradiente sobrevive, no es composición.
est_ciudad = cui["MOMENTOCURSOVIDA"].value_counts(normalize=True)
tasas = cui.pivot_table(index="NOMBRE_LOCALIDAD", columns="MOMENTOCURSOVIDA",
                        values="CUID_SEXO", aggfunc=lambda s: (s == "Mujer").mean())
n_celda = cui.pivot_table(index="NOMBRE_LOCALIDAD", columns="MOMENTOCURSOVIDA",
                          values="CUID_SEXO", aggfunc="size")
tasas = tasas.where(n_celda >= N_MIN_CELDA)          # ignora celdas minúsculas
ind_loc["IFC_std"] = (tasas.mul(est_ciudad, axis=1).sum(axis=1) /
                      tasas.notna().mul(est_ciudad, axis=1).sum(axis=1)) * 100

# ---------- Marcar lo que no tiene base suficiente --------------------------
ind_loc["base_suficiente"] = ind_loc["n_cuidado"] >= N_MIN
excluidas = ind_loc[~ind_loc["base_suficiente"]]

ind_loc.round(2).to_csv(TABLAS / "indicadores_localidad.csv", sep=SEP)

print(f"[7] Umbral mínimo: {N_MIN} cuidadores por localidad\n")
if len(excluidas):
    print(f"  Localidades con base insuficiente ({len(excluidas)}), marcadas y excluidas:")
    for n, r in excluidas.iterrows():
        print(f"      {n:<20} n = {int(r['n_cuidado']):>4}   "
              f"(adultos: {int(r['n_adultos'])})")

ok = ind_loc[ind_loc["base_suficiente"]].sort_values("IFC_std", ascending=False)
print(f"\n  {len(ok)} localidades en el análisis territorial:\n")
display(ok[["n_cuidado", "IFC", "IFC_std", "CCM", "CSR", "DC"]].round(2))
```

### R2.2 · ¿El gradiente es composición o es real? ¿Con qué se asocia?


```python
g = ind_loc[ind_loc["base_suficiente"]].dropna(subset=["IFC_std"])

# --- ¿Es composición etaria? ------------------------------------------------
rho_std   = spearman(g["IFC"], g["IFC_std"])
amp_cruda = g["IFC"].max() - g["IFC"].min()
amp_std   = g["IFC_std"].max() - g["IFC_std"].min()
print("¿EL GRADIENTE ES COMPOSICIÓN ETARIA?")
print(f"  Correlación cruda vs estandarizada : {rho_std:.3f}")
print(f"  Amplitud cruda {amp_cruda:.1f} pp -> estandarizada {amp_std:.1f} pp")
print("  " + ("El gradiente SOBREVIVE: no lo produce la composición."
              if rho_std > 0.9 and amp_std > amp_cruda*0.7 else
              "⚠ Buena parte del gradiente es composición: repórtelo."))

# --- ¿Con qué se asocia? ----------------------------------------------------
eco = base_dep.groupby("NOMBRE_LOCALIDAD").apply(lambda x: pd.Series({
    "pct_superior": x["NIVELEDUCATIVO"].isin(
        ["Educación Superior Pregrado", "Educación Superior Posgrado"]).mean() * 100,
    "pct_ingreso_trabajo": (x["FUENTEINGRESOS"] == "Del Trabajo").mean() * 100,
}), include_groups=False)
g2 = g.join(eco)
print("\n¿CON QUÉ SE ASOCIA?")
for v in ["pct_superior", "pct_ingreso_trabajo"]:
    print(f"  IFC* vs {v:20s}: Spearman {spearman(g2['IFC_std'], g2[v]):+.3f}")
print("  Signo negativo = donde hay menos capacidad económica, la carga")
print("  de cuidado recae más sobre mujeres.")
g2.round(2).to_csv(TABLAS / "gradiente_socioeconomico.csv", sep=SEP)

fig, ax = plt.subplots(figsize=(9, 5.4))
ax.scatter(g2["pct_superior"], g2["IFC_std"], s=g2["n_cuidado"]/12,
           color=ROSA, alpha=.75, zorder=3, edgecolor="white", linewidth=1.2)
b, a = np.polyfit(g2["pct_superior"], g2["IFC_std"], 1)
xs = np.linspace(g2["pct_superior"].min()-1, g2["pct_superior"].max()+1, 50)
ax.plot(xs, a + b*xs, color=AZUL, lw=1.6, ls="--", zorder=2)
for k, r in g2.iterrows():
    ax.annotate(k, (r["pct_superior"], r["IFC_std"]), fontsize=7.5,
                xytext=(6,-3), textcoords="offset points", color="#333333")
ax.text(.98,.95, f"Spearman = {spearman(g2['IFC_std'], g2['pct_superior']):+.2f}",
        transform=ax.transAxes, ha="right", fontsize=10, color=AZUL, fontweight="bold")
ax.set_xlabel("% de personas certificadas con educación superior  (proxy de posición económica)")
ax.set_ylabel("Feminización estandarizada (%)")
ax.set_title("Donde hay menos capacidad económica, el cuidado recae más en mujeres",
             fontsize=12.5, fontweight="bold", color=AZUL, loc="left")
ax.grid(alpha=.22, zorder=0)
fig.tight_layout(); fig.savefig(FIGURAS / "08_gradiente_socioeconomico.png", dpi=200, bbox_inches="tight")
plt.show()
```

## R3 · ¿Dónde se concentra la carga?

*Pregunta principal, primera parte.*



| Medida | Qué responde |
|---|---|
| **Volumen absoluto** | Dónde están las personas. Manda el tamaño de la localidad. |
| **TCC** — tasa de carga por 1.000 habitantes | Dónde pesa más el fenómeno respecto a su población. |
| **DC** — % de certificados que requieren ayuda | Qué tan severa es la discapacidad certificada allí. |

```python
carga = (panel_int.groupby("COD_LOC")
         .agg(requieren_ayuda=("requieren_ayuda","sum"),
              TCC=("TCC","mean"), POBLACION=("POBLACION","max"))
         .reset_index())
carga["NOMBRE_LOCALIDAD"] = carga["COD_LOC"].map(LOCALIDADES)
carga = carga.merge(ind_loc[["DC","n_cuidado","base_suficiente"]].reset_index(),
                    on="NOMBRE_LOCALIDAD", how="left")
carga["pct_del_total"] = carga["requieren_ayuda"] / carga["requieren_ayuda"].sum() * 100

print("POR VOLUMEN ABSOLUTO — dónde están las personas")
display(carga.nlargest(6, "requieren_ayuda")[
    ["NOMBRE_LOCALIDAD","requieren_ayuda","pct_del_total","TCC","DC"]].round(2))

print("\nPOR INTENSIDAD (TCC) — dónde pesa más respecto a su población")
display(carga.nlargest(6, "TCC")[
    ["NOMBRE_LOCALIDAD","requieren_ayuda","POBLACION","TCC","DC"]].round(2))

top5 = carga.nlargest(5, "requieren_ayuda")
print(f"\n  Las 5 localidades con más casos concentran el "
      f"{top5['pct_del_total'].sum():.1f} % de toda la carga de la ciudad.")
rho = spearman(carga["requieren_ayuda"], carga["TCC"])
print(f"  Correlación entre volumen e intensidad: {rho:+.3f}")
print("  " + ("Las dos lecturas coinciden." if rho > .8 else
      "Las dos lecturas NO coinciden: reportar ambas y decir cuál se usó para priorizar."))
carga.round(3).to_csv(TABLAS / "concentracion_carga.csv", sep=SEP, index=False)
```

## R5 · ¿Qué tan alineada está la oferta? ¿Hay territorios con alta demanda y baja oferta?

*Pregunta principal, tercera parte, y subpregunta 3.*

$$ICOD_i = z[OC_i] - z[TCC_i]$$

La demanda se mide con la tasa de carga por mil habitantes; la oferta, con servicios de cuidado por cien mil habitantes. Cada lado se lee en su propia unidad y no hay ponderaciones que justificar.

| Valor | Situación | Recomendación |
|---|---|---|
| Muy negativo | Alta carga, baja oferta | **Prioridad de expansión** |
| Cercano a cero | Oferta proporcional | Sostener y monitorear |
| Muy positivo | Oferta alta frente a la carga registrada | Revisar si la carga está **subregistrada** antes de concluir sobreoferta |



```python
# --- Guarda: la oferta es opcional -----------------------------------------
if "mz" not in dir() or mz is None or len(mz) == 0:
    print("  ⚠ No hay datos de Manzanas del Cuidado.")
    print("    Ejecute la sección A.7.4 antes de esta celda.")
    print("    El resto del análisis no depende de esto: puede continuar.")
    HAY_OFERTA = False
else:
    HAY_OFERTA = True

if HAY_OFERTA:
    acum_int = (panel_int.groupby("COD_LOC")
                .agg(IFC=("IFC", "mean"), TCC=("TCC", "mean"),
                     requieren_ayuda=("requieren_ayuda", "sum"),
                     POBLACION=("POBLACION", "max"))
                .reset_index())
    acum_int["NOMBRE_LOCALIDAD"] = acum_int["COD_LOC"].map(LOCALIDADES)

    # Traer la feminización estandarizada y el filtro de base suficiente
    acum_int = acum_int.merge(
        ind_loc[["IFC_std", "n_cuidado", "base_suficiente"]].reset_index()
            .rename(columns={"NOMBRE_LOCALIDAD": "NOMBRE_LOCALIDAD"}),
        on="NOMBRE_LOCALIDAD", how="left")

    cols_mz = [c for c in ["COD_LOC", "MANZANAS", "SERVICIOS",
                           "Para quienes requieren Cuidado"] if c in mz.columns]
    acum_int = vincular(acum_int, mz[cols_mz], sobre=("COD_LOC",),
                        nombre_izq="panel acumulado", nombre_der="manzanas")

    # Ausencia informativa: sin manzana = oferta cero
    for c in cols_mz[1:]:
        acum_int[c] = acum_int[c].fillna(0)

    acum_int["OC"] = acum_int["SERVICIOS"] / acum_int["POBLACION"] * 100_000
    acum_int["ICOD"] = z(acum_int["OC"]) - z(acum_int["TCC"])
    acum_int["clasificacion"] = pd.cut(
        acum_int["ICOD"], [-99, -0.5, 0.5, 99],
        labels=["Prioridad de expansión", "Sostener y monitorear",
                "Revisar cobertura o subregistro"])

    acum_int.round(3).to_csv(TABLAS / "correspondencia_oferta_demanda.csv", sep=SEP, index=False)

    n_bajo = (~acum_int["base_suficiente"].fillna(False)).sum()
    if n_bajo:
        print(f"\n  ⚠ {n_bajo} localidad(es) con menos de {N_MIN} cuidadores: "
              "se muestran marcadas, no se usan para conclusiones.")

    print("\n  Demanda = TCC (por 1.000 hab.) · Oferta = OC (por 100.000 hab.)\n")
    vista = acum_int.sort_values("ICOD")[
        ["NOMBRE_LOCALIDAD", "requieren_ayuda", "TCC", "IFC_std",
         "MANZANAS", "SERVICIOS", "OC", "ICOD", "clasificacion", "base_suficiente"]]
    display(vista.round(2))
```

```python
PLANTILLA_TABLERO = r'''<!DOCTYPE html>
<html lang="es">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>Quién paga el cuidado y quién lo absorbe · Tablero</title>
<style>
:root{
  --azul:#1F3864; --azul2:#2E4A7D; --rosa:#C0504D; --rosa2:#D98C8A;
  --gris:#8A8F98; --linea:#E3E7ED; --fondo:#F6F7F9; --panel:#FFFFFF; --texto:#22262C;
}
*{box-sizing:border-box}
body{margin:0;background:var(--fondo);color:var(--texto);
     font-family:"Segoe UI",system-ui,-apple-system,Roboto,Helvetica,Arial,sans-serif;font-size:14px}
header{background:var(--azul);color:#fff;padding:22px 28px 46px}
header .eyebrow{font-size:11px;letter-spacing:2.4px;opacity:.72;font-weight:600}
header h1{margin:8px 0 2px;font-size:26px;font-weight:700;letter-spacing:-.3px}
header p{margin:0;opacity:.82;font-size:13.5px}
header p.aut{margin-top:9px;opacity:.62;font-size:12px;letter-spacing:.2px}
.wrap{max-width:1280px;margin:0 auto;padding:0 20px 48px}
.kpis{display:grid;grid-template-columns:repeat(auto-fit,minmax(168px,1fr));gap:12px;margin:-26px 0 22px}
.kpi{background:var(--panel);border:1px solid var(--linea);border-radius:10px;padding:14px 16px;
     box-shadow:0 2px 10px rgba(31,56,100,.07)}
.kpi .v{font-size:26px;font-weight:700;color:var(--azul);line-height:1.1}
.kpi .l{font-size:11.5px;color:var(--gris);margin-top:5px;line-height:1.35}
.kpi.acento .v{color:var(--rosa)}
.grid{display:grid;gap:16px;grid-template-columns:repeat(12,1fr)}
.card{background:var(--panel);border:1px solid var(--linea);border-radius:10px;padding:16px 18px 14px;
      box-shadow:0 1px 5px rgba(31,56,100,.05)}
.c12{grid-column:span 12}.c7{grid-column:span 7}.c5{grid-column:span 5}
.c6{grid-column:span 6}.c8{grid-column:span 8}.c4{grid-column:span 4}
@media(max-width:980px){.c7,.c5,.c6,.c8,.c4{grid-column:span 12}}
.card h2{margin:0 0 3px;font-size:15.5px;color:var(--azul);font-weight:700}
.card .q{font-size:11.5px;color:var(--rosa);font-weight:700;letter-spacing:.4px;text-transform:uppercase;margin-bottom:7px}
.card .sub{font-size:12.5px;color:var(--gris);margin:0 0 12px;line-height:1.45}
.sel{display:flex;gap:6px;flex-wrap:wrap;margin-bottom:10px}
.sel button{border:1px solid var(--linea);background:#fff;border-radius:20px;padding:5px 12px;
  font-size:12px;cursor:pointer;color:var(--azul);font-family:inherit;transition:.15s}
.sel button:hover{border-color:var(--azul2)}
.sel button.on{background:var(--azul);color:#fff;border-color:var(--azul)}
svg{display:block;width:100%;overflow:visible}
.bar{cursor:pointer;transition:opacity .15s}
.bar:hover{opacity:.78}
.lbl{font-size:11px;fill:var(--texto)}
.lbl.sm{font-size:10px;fill:var(--gris)}
.val{font-size:10.5px;fill:var(--texto);font-weight:600}
.ax{stroke:var(--linea);stroke-width:1}
.axt{font-size:10px;fill:var(--gris)}
.pt{cursor:pointer;transition:.15s}
.pt:hover{stroke:var(--azul);stroke-width:2}
.sel-loc{stroke:var(--azul);stroke-width:2.5}
.leyenda{display:flex;gap:16px;font-size:11.5px;color:var(--gris);margin-top:8px;flex-wrap:wrap}
.leyenda i{display:inline-block;width:11px;height:11px;border-radius:2px;margin-right:5px;vertical-align:-1px}
table{width:100%;border-collapse:collapse;font-size:12.5px}
th{text-align:left;font-weight:600;color:#fff;background:var(--azul);padding:7px 9px;font-size:11.5px}
td{padding:6px 9px;border-bottom:1px solid var(--linea)}
tr.baja td{color:var(--gris);font-style:italic}
tr:hover td{background:#F0F4FA;cursor:pointer}
.nota{font-size:11.5px;color:var(--gris);margin-top:10px;line-height:1.5;
      border-left:3px solid var(--linea);padding-left:10px}
.alerta{background:#FFF6E8;border-left:3px solid #C9761E;padding:10px 12px;border-radius:0 6px 6px 0;
        font-size:12px;line-height:1.5;margin-top:10px}
.detalle{background:#F0F4FA;border-radius:8px;padding:12px 14px;font-size:12.5px;line-height:1.6}
.detalle b{color:var(--azul)}
footer{background:#fff;border-top:1px solid var(--linea);padding:20px 28px;font-size:11.5px;
       color:var(--gris);line-height:1.6}
footer b{color:var(--azul)}
</style>
</head>
<body>
<header>
  <div class="eyebrow">DATAJAM MULTIUNIVERSITARIO 2026 · EDICIÓN 3</div>
  <h1>Quién paga el cuidado y quién lo absorbe</h1>
  <p>Cuidado no remunerado derivado de la discapacidad en Bogotá D.C. · Vigencias 2020–2026</p>
  <p class="aut">María Clareth Méndez Ramos · Jose Carlos Arroyo Cantero · Valentina Terán Barreto</p>
</header>

<div class="wrap">
  <div class="kpis" id="kpis"></div>

  <div class="grid">
    <div class="card c7">
      <div class="q">R1 · ¿Con qué grado de feminización?</div>
      <h2>El cuidado no se reparte en ninguna etapa de la vida</h2>
      <p class="sub">Sexo de la persona cuidadora según la etapa de vida de quien recibe el cuidado.</p>
      <svg id="g_curso" viewBox="0 0 560 250"></svg>
      <div class="leyenda"><span><i style="background:var(--rosa)"></i>Cuida una mujer</span>
        <span><i style="background:var(--azul)"></i>Cuida un hombre</span></div>
    </div>

    <div class="card c5">
      <div class="q">R1 · Mismo parentesco, distinto sexo</div>
      <h2>La madre aparece 14 veces por cada padre</h2>
      <p class="sub">Porcentaje del total de personas cuidadoras.</p>
      <svg id="g_par" viewBox="0 0 420 250"></svg>
    </div>

    <div class="card c7">
      <div class="q">R2 · ¿En qué localidades es mayor?</div>
      <h2>Ranking territorial</h2>
      <p class="sub">Haga clic en una localidad para resaltarla en todos los paneles.</p>
      <div class="sel" id="sel_metrica"></div>
      <svg id="g_rank" viewBox="0 0 560 430"></svg>
    </div>

    <div class="card c5">
      <div class="q">R2 · ¿Con qué se asocia?</div>
      <h2>El gradiente es socioeconómico</h2>
      <p class="sub">Feminización estandarizada por estructura etaria, frente al nivel educativo del territorio.</p>
      <svg id="g_grad" viewBox="0 0 420 300"></svg>
      <div class="nota" id="nota_grad"></div>
    </div>

    <div class="card c6">
      <div class="q">R5 · ¿Está la oferta donde está la carga?</div>
      <h2>Oferta del Sistema Distrital de Cuidado frente a la demanda</h2>
      <p class="sub">Servicios de las Manzanas del Cuidado frente a personas con cuidador familiar identificado.</p>
      <svg id="g_of" viewBox="0 0 460 300"></svg>
      <div class="alerta" id="alerta_of"></div>
    </div>

    <div class="card c6">
      <div class="q">R4 · ¿Cómo evolucionó 2020–2026?</div>
      <h2>La feminización cede, pero solo en la adultez</h2>
      <p class="sub">Porcentaje de cuidadores familiares que son mujeres, por vigencia.</p>
      <svg id="g_serie" viewBox="0 0 460 240"></svg>
      <div class="nota">La vigencia 2020 tiene muy pocos registros y no es comparable.
        De la caída total, cerca de una cuarta parte se explica por el cambio de composición
        etaria del registro; el resto es cambio real y se concentra en la adultez.</div>
    </div>

    <div class="card c8">
      <div class="q">Detalle</div>
      <h2>Indicadores por localidad</h2>
      <p class="sub">IFC*: feminización estandarizada · CCM: cuidado materno · CSR: adultos aún cuidados por su madre · DC: certificados que requieren ayuda.</p>
      <div style="max-height:330px;overflow:auto"><table id="tabla"></table></div>
      <div class="nota">Las localidades en gris tienen menos de 300 cuidadores identificados:
        se muestran por transparencia pero no sustentan conclusiones.</div>
    </div>

    <div class="card c4">
      <div class="q">Ficha</div>
      <h2 id="det_titulo">Seleccione una localidad</h2>
      <div class="detalle" id="detalle">Haga clic en cualquier barra, punto o fila de la tabla.</div>
    </div>
  </div>
</div>

<footer>
  <b>Fuentes.</b> Población con discapacidad certificada en Bogotá D.C. — Secretaría Distrital de Salud,
  Observatorio de Salud de Bogotá · Manzanas del Cuidado (Equipamiento Ancla y Servicios de la Manzana) —
  Secretaría Distrital de la Mujer, vía servicio REST de IDECA, corte 16/12/2025.<br>
  <b>Advertencias.</b> El registro mide <i>certificaciones</i>, no prevalencia: las diferencias territoriales
  mezclan incidencia real con capacidad de acceder al trámite. Los indicadores de cuidado remunerado
  requieren validación externa. El diseño es observacional: las asociaciones no son efectos causales.
  El nivel educativo se usa como indicador aproximado de la posición económica del territorio.<br>
  <b>Equipo.</b> María Clareth Méndez Ramos · Jose Carlos Arroyo Cantero · Valentina Terán Barreto<br>
  <b>Generado por</b> el cuaderno de análisis del ejercicio. Formato HTML autocontenido, sin dependencias externas.
</footer>

<script>
const D = /*DATOS*/{}/*FIN*/;
const AZUL="#1F3864", ROSA="#C0504D", GRIS="#8A8F98", ROSA2="#D98C8A";
let selLoc = null;
const ok = D.locs.filter(l=>l.ok);
const N = (x,d=1)=> x==null||isNaN(x) ? "—" : Number(x).toLocaleString("es-CO",{minimumFractionDigits:d,maximumFractionDigits:d});
const E = (t,a={},h="")=>{const s=`<${t} ${Object.entries(a).map(([k,v])=>`${k}="${v}"`).join(" ")}>${h}</${t}>`;return s};

/* ---------- KPIs ---------- */
const k = D.kpi;
document.getElementById("kpis").innerHTML = [
 [`${N(k.ifc,1)} %`, "de los cuidadores familiares son mujeres", 1],
 [`${N(k.razon,2)} a 1`, "razón mujer : hombre entre quienes cuidan", 1],
 [k.mujeres.toLocaleString("es-CO"), "mujeres cuidadoras identificadas", 0],
 [k.madre.toLocaleString("es-CO"), "personas cuidadas por su madre", 0],
 [`${N(k.csr,1)} %`, `adultos aún cuidados por su madre (${k.csr_n.toLocaleString("es-CO")} casos)`, 1],
 [`${k.manzanas} / ${k.servicios}`, "manzanas y servicios de cuidado en la ciudad", 0],
].map(([v,l,ac])=>`<div class="kpi ${ac?'acento':''}"><div class="v">${v}</div><div class="l">${l}</div></div>`).join("");

/* ---------- Curso de vida ---------- */
(()=>{const d=D.mujeresPorEtapa, W=560,H=250,ml=118,mr=52,mt=8,bh=(H-mt-14)/d.length;
 let s="";
 d.forEach((r,i)=>{const y=mt+i*bh, h=bh*0.62, w=(W-ml-mr)*r.mujer/100;
  s+=E("rect",{x:ml,y:y,width:w,height:h,fill:ROSA,rx:2});
  s+=E("rect",{x:ml+w,y:y,width:(W-ml-mr)-w,height:h,fill:AZUL,rx:2});
  s+=E("text",{x:ml-8,y:y+h/2+4,"text-anchor":"end",class:"lbl"},r.etapa);
  s+=E("text",{x:ml+w-6,y:y+h/2+4,"text-anchor":"end",class:"val",fill:"#fff"},N(r.mujer,0)+" %");
 });
 document.getElementById("g_curso").innerHTML=s;})();

/* ---------- Parentesco ---------- */
(()=>{const cv=D.cursoVida, tot=D.kpi.total;
 // porcentajes globales por parentesco
 const p=[["Madre / Padre",D.kpi.madre/tot*100,null]];
 const g=D.parentescoGlobal;
 const W=420,H=250,ml=104,mr=46,mt=10,gh=(H-mt-10)/g.length;
 let s="";
 g.forEach((r,i)=>{const y=mt+i*gh, h=gh*0.30, mx=Math.max(...g.map(x=>Math.max(x.mujer,x.hombre)));
  const w1=(W-ml-mr)*r.mujer/mx, w2=(W-ml-mr)*r.hombre/mx;
  s+=E("rect",{x:ml,y:y,width:w1,height:h,fill:ROSA,rx:2});
  s+=E("rect",{x:ml,y:y+h+3,width:w2,height:h,fill:AZUL,rx:2});
  s+=E("text",{x:ml+w1+5,y:y+h-1,class:"val"},N(r.mujer,1)+" %");
  s+=E("text",{x:ml+w2+5,y:y+2*h+2,class:"val"},N(r.hombre,1)+" %");
  s+=E("text",{x:ml-8,y:y+h+2,"text-anchor":"end",class:"lbl"},r.par);
 });
 document.getElementById("g_par").innerHTML=s;})();

/* ---------- Ranking ---------- */
const METRICAS=[["ifc_std","Feminización (IFC*)","%"],["csr","Sin relevo (CSR)","%"],
                ["dc","Demanda (DC)","%"],["n","Volumen","casos"]];
let metrica="ifc_std";
document.getElementById("sel_metrica").innerHTML =
  METRICAS.map(([k2,l])=>`<button data-m="${k2}" class="${k2==metrica?'on':''}">${l}</button>`).join("");
document.querySelectorAll("#sel_metrica button").forEach(b=>b.onclick=()=>{
  metrica=b.dataset.m;
  document.querySelectorAll("#sel_metrica button").forEach(x=>x.classList.toggle("on",x===b));
  dibujarRank();});

function dibujarRank(){
  const m=METRICAS.find(x=>x[0]===metrica);
  const d=[...ok].filter(l=>l[metrica]!=null).sort((a,b)=>b[metrica]-a[metrica]);
  const W=560,H=430,ml=124,mr=54,mt=6,bh=(H-mt-10)/d.length, mx=Math.max(...d.map(l=>l[metrica]));
  const mn = metrica==="n" ? 0 : Math.min(...d.map(l=>l[metrica]))*0.94;
  let s="";
  d.forEach((l,i)=>{const y=mt+i*bh,h=bh*0.66;
    const w=(W-ml-mr)*((l[metrica]-mn)/(mx-mn||1))*0.96+4;
    const col = selLoc===l.loc ? AZUL : ROSA;
    s+=E("rect",{x:ml,y:y,width:w,height:h,fill:col,rx:2,class:"bar","data-loc":l.loc});
    s+=E("text",{x:ml-8,y:y+h/2+3.5,"text-anchor":"end",class:"lbl"+(selLoc===l.loc?"":""),
                 "font-weight":selLoc===l.loc?"700":"400"},l.loc);
    s+=E("text",{x:ml+w+6,y:y+h/2+3.5,class:"val"}, m[2]==="%"?N(l[metrica],1)+" %":l[metrica].toLocaleString("es-CO"));
  });
  document.getElementById("g_rank").innerHTML=s;
  document.querySelectorAll("#g_rank .bar").forEach(b=>b.onclick=()=>elegir(b.dataset.loc));
}

/* ---------- Gradiente ---------- */
function dibujarGrad(){
  const d=ok.filter(l=>l.ifc_std!=null&&l.sup!=null);
  const W=420,H=300,ml=50,mr=14,mt=12,mb=40;
  const xs=d.map(l=>l.sup), ys=d.map(l=>l.ifc_std);
  const x0=Math.min(...xs)-2,x1=Math.max(...xs)+2,y0=Math.min(...ys)-.6,y1=Math.max(...ys)+.6;
  const X=v=>ml+(W-ml-mr)*(v-x0)/(x1-x0), Y=v=>H-mb-(H-mt-mb)*(v-y0)/(y1-y0);
  const n=d.length, sx=xs.reduce((a,b)=>a+b,0), sy=ys.reduce((a,b)=>a+b,0);
  const sxy=d.reduce((a,l)=>a+l.sup*l.ifc_std,0), sxx=xs.reduce((a,b)=>a+b*b,0);
  const b1=(n*sxy-sx*sy)/(n*sxx-sx*sx), b0=(sy-b1*sx)/n;
  let s="";
  s+=E("line",{x1:ml,y1:H-mb,x2:W-mr,y2:H-mb,class:"ax"});
  s+=E("line",{x1:ml,y1:mt,x2:ml,y2:H-mb,class:"ax"});
  for(let i=0;i<=4;i++){const v=y0+(y1-y0)*i/4;
    s+=E("text",{x:ml-6,y:Y(v)+3,"text-anchor":"end",class:"axt"},N(v,0));}
  for(let i=0;i<=4;i++){const v=x0+(x1-x0)*i/4;
    s+=E("text",{x:X(v),y:H-mb+15,"text-anchor":"middle",class:"axt"},N(v,0)+"%");}
  s+=E("line",{x1:X(x0),y1:Y(b0+b1*x0),x2:X(x1),y2:Y(b0+b1*x1),stroke:AZUL,"stroke-width":1.4,"stroke-dasharray":"5 4"});
  d.forEach(l=>{const r=Math.max(4,Math.sqrt(l.n)/9);
    s+=E("circle",{cx:X(l.sup),cy:Y(l.ifc_std),r:r,fill:ROSA,"fill-opacity":.75,
      stroke:selLoc===l.loc?AZUL:"#fff","stroke-width":selLoc===l.loc?2.5:1.2,class:"pt","data-loc":l.loc});});
  s+=E("text",{x:(W+ml)/2,y:H-8,"text-anchor":"middle",class:"axt"},"% con educación superior");
  const cy1=(mt+(H-mb))/2;
  s+=E("text",{x:11,y:cy1,class:"axt","text-anchor":"middle",
               transform:`rotate(-90 11 ${cy1})`},"IFC* (%)");
  document.getElementById("g_grad").innerHTML=s;
  document.querySelectorAll("#g_grad .pt").forEach(p=>p.onclick=()=>elegir(p.dataset.loc));
  // rho de Spearman
  const rk=a=>{const s2=[...a].sort((x,y)=>x-y);return a.map(v=>s2.indexOf(v)+1)};
  const rx=rk(xs), ry=rk(ys), mrx=rx.reduce((a,b)=>a+b,0)/n, mry=ry.reduce((a,b)=>a+b,0)/n;
  let num=0,dx=0,dy=0; for(let i=0;i<n;i++){num+=(rx[i]-mrx)*(ry[i]-mry);dx+=(rx[i]-mrx)**2;dy+=(ry[i]-mry)**2;}
  const rho=num/Math.sqrt(dx*dy);
  document.getElementById("nota_grad").innerHTML =
    `Correlación de rangos <b>Spearman = ${N(rho,2)}</b>. El tamaño del círculo es proporcional al
     número de cuidadores. La estandarización etaria confirma que el gradiente no lo produce
     la composición de edades de cada registro local.`;
}

/* ---------- Oferta vs demanda ---------- */
function dibujarOferta(){
  const d=ok;
  const W=460,H=300,ml=52,mr=16,mt=12,mb=40;
  const x1=Math.max(...d.map(l=>l.n))*1.06, y1=Math.max(...d.map(l=>l.servicios))*1.12;
  const X=v=>ml+(W-ml-mr)*v/x1, Y=v=>H-mb-(H-mt-mb)*v/y1;
  let s="";
  s+=E("line",{x1:ml,y1:H-mb,x2:W-mr,y2:H-mb,class:"ax"});
  s+=E("line",{x1:ml,y1:mt,x2:ml,y2:H-mb,class:"ax"});
  const kk=d.reduce((a,l)=>a+l.servicios,0)/d.reduce((a,l)=>a+l.n,0);
  s+=E("line",{x1:X(0),y1:Y(0),x2:X(x1),y2:Y(kk*x1),stroke:GRIS,"stroke-width":1.2,"stroke-dasharray":"5 4"});
  s+=E("text",{x:X(x1)-4,y:Y(kk*x1)-6,"text-anchor":"end",class:"axt"},"oferta proporcional a la demanda");
  // Puntos
  d.forEach(l=>{const bajo = l.servicios < kk*l.n*0.6;
    s+=E("circle",{cx:X(l.n),cy:Y(l.servicios),r:6,fill:bajo?ROSA:AZUL,"fill-opacity":.8,
      stroke:selLoc===l.loc?AZUL:"#fff","stroke-width":selLoc===l.loc?2.5:1.2,class:"pt","data-loc":l.loc});});

  // Etiquetas con separación vertical: si dos quedan a menos de 12 px se
  // desplaza la siguiente, y si la etiqueta se saldría por la derecha se
  // coloca a la izquierda del punto.
  const et = d.filter(l=>l.servicios < kk*l.n*0.6 || selLoc===l.loc)
              .map(l=>({loc:l.loc, x:X(l.n), y:Y(l.servicios)}))
              .sort((a,b)=>a.y-b.y);
  let ultimo = -99;
  et.forEach(e=>{
    let y = e.y + 3.5;
    if(y - ultimo < 12) y = ultimo + 12;
    ultimo = y;
    const derecha = e.x + 9 + e.loc.length*5.4 < W - mr + 30;
    s+=E("text",{x: derecha ? e.x+9 : e.x-9, y:y,
                 "text-anchor": derecha ? "start" : "end",
                 class:"lbl sm"}, e.loc);
  });
  for(let i=0;i<=3;i++){const v=y1*i/3; s+=E("text",{x:ml-6,y:Y(v)+3,"text-anchor":"end",class:"axt"},Math.round(v));}
  for(let i=0;i<=3;i++){const v=x1*i/3; s+=E("text",{x:X(v),y:H-mb+15,"text-anchor":"middle",class:"axt"},Math.round(v).toLocaleString("es-CO"));}
  s+=E("text",{x:(W+ml)/2,y:H-8,"text-anchor":"middle",class:"axt"},"personas con cuidador familiar identificado");
  const cy2=(mt+(H-mb))/2;
  s+=E("text",{x:11,y:cy2,class:"axt","text-anchor":"middle",
               transform:`rotate(-90 11 ${cy2})`},"servicios de cuidado");
  document.getElementById("g_of").innerHTML=s;
  document.querySelectorAll("#g_of .pt").forEach(p=>p.onclick=()=>elegir(p.dataset.loc));
  const bajos=d.filter(l=>l.servicios<kk*l.n*0.6).sort((a,b)=>(a.servicios/a.n)-(b.servicios/b.n));
  document.getElementById("alerta_of").innerHTML =
    `<b>Territorios con alta demanda y baja oferta:</b> ${bajos.map(l=>l.loc).join(" · ")}.
     Son los candidatos a priorización para la expansión del Sistema Distrital de Cuidado.
     La línea punteada marca la proporción promedio de la ciudad.`;
}

/* ---------- Serie ---------- */
(()=>{const d=D.serie.filter(r=>r.n>=1000);
 const W=460,H=240,ml=48,mr=16,mt=14,mb=34;
 const y0=Math.min(...d.map(r=>r.ifc))-2, y1=Math.max(...d.map(r=>r.ifc))+2;
 const X=i=>ml+(W-ml-mr)*i/(d.length-1), Y=v=>H-mb-(H-mt-mb)*(v-y0)/(y1-y0);
 let s="";
 s+=E("line",{x1:ml,y1:H-mb,x2:W-mr,y2:H-mb,class:"ax"});
 for(let i=0;i<=3;i++){const v=y0+(y1-y0)*i/3;
   s+=E("line",{x1:ml,y1:Y(v),x2:W-mr,y2:Y(v),stroke:"#EFF2F6"});
   s+=E("text",{x:ml-6,y:Y(v)+3,"text-anchor":"end",class:"axt"},N(v,0));}
 s+=E("polyline",{points:d.map((r,i)=>`${X(i)},${Y(r.ifc)}`).join(" "),fill:"none",stroke:ROSA,"stroke-width":2.4});
 d.forEach((r,i)=>{s+=E("circle",{cx:X(i),cy:Y(r.ifc),r:4.5,fill:ROSA,stroke:"#fff","stroke-width":1.6});
   s+=E("text",{x:X(i),y:Y(r.ifc)-11,"text-anchor":"middle",class:"val"},N(r.ifc,1));
   s+=E("text",{x:X(i),y:H-mb+15,"text-anchor":"middle",class:"axt"},r.anio);});
 const cy3=(mt+(H-mb))/2;
 s+=E("text",{x:11,y:cy3,class:"axt","text-anchor":"middle",
              transform:`rotate(-90 11 ${cy3})`},"% mujeres");
 document.getElementById("g_serie").innerHTML=s;})();

/* ---------- Tabla ---------- */
function dibujarTabla(){
  const d=[...D.locs].sort((a,b)=>b.n-a.n);
  let s="<tr><th>Localidad</th><th>Cuidadores</th><th>IFC*</th><th>CCM</th><th>CSR</th><th>DC</th><th>Manzanas</th><th>Servicios</th></tr>";
  d.forEach(l=>{s+=`<tr class="${l.ok?'':'baja'}" data-loc="${l.loc}">
    <td>${l.loc}</td><td>${l.n.toLocaleString("es-CO")}</td>
    <td>${l.ifc_std==null?"—":N(l.ifc_std,1)+" %"}</td><td>${N(l.ccm,1)} %</td>
    <td>${N(l.csr,1)} %</td><td>${N(l.dc,1)} %</td>
    <td>${l.manzanas}</td><td>${l.servicios}</td></tr>`;});
  document.getElementById("tabla").innerHTML=s;
  document.querySelectorAll("#tabla tr[data-loc]").forEach(t=>t.onclick=()=>elegir(t.dataset.loc));
}

/* ---------- Selección ---------- */
function elegir(loc){
  selLoc = selLoc===loc ? null : loc;
  dibujarRank(); dibujarGrad(); dibujarOferta();
  const l=D.locs.find(x=>x.loc===selLoc);
  const t=document.getElementById("det_titulo"), c=document.getElementById("detalle");
  if(!l){t.textContent="Seleccione una localidad";
    c.innerHTML="Haga clic en cualquier barra, punto o fila de la tabla.";return;}
  t.textContent=l.loc;
  const kk=ok.reduce((a,x)=>a+x.servicios,0)/ok.reduce((a,x)=>a+x.n,0);
  const esp=Math.round(kk*l.n);
  c.innerHTML=`
   <b>${l.n.toLocaleString("es-CO")}</b> personas con cuidador familiar identificado<br>
   <b>${l.cert.toLocaleString("es-CO")}</b> certificaciones · <b>${N(l.dc,1)} %</b> requiere ayuda<br><br>
   Feminización estandarizada · <b>${l.ifc_std==null?"—":N(l.ifc_std,1)+" %"}</b><br>
   Cuidado materno · <b>${N(l.ccm,1)} %</b><br>
   Sin relevo (adultos) · <b>${N(l.csr,1)} %</b> de ${l.nad}<br>
   Educación superior · <b>${N(l.sup,1)} %</b><br><br>
   Manzanas del Cuidado · <b>${l.manzanas}</b><br>
   Servicios · <b>${l.servicios}</b> (esperados ${esp} si la oferta fuera proporcional)<br>
   Servicios «para quienes requieren cuidado» · <b>${l.serv_req}</b>
   ${l.ok?"":"<br><br><i>Base insuficiente: menos de 300 cuidadores. No sustenta conclusiones.</i>"}`;
}

dibujarRank(); dibujarGrad(); dibujarOferta(); dibujarTabla();
</script>
</body>
</html>
'''
print('Plantilla cargada:', len(PLANTILLA_TABLERO)//1024, 'KB')
```
